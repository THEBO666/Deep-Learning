```python
import numpy as np


class MulLayer:
    def __init__(self):
        self.x = None
        self.y = None

    def forward(self, x, y):
        self.x = x
        self.y = y
        out = x * y

        return out

    def backward(self, x, y):
        dy = x
        dx = y
        return dx, dy


class AddLayer:
    def __init__(self):
        pass

    def forward(self, x, y):
        out = x + y
        return out

    def backward(self, dout):
        dx = dout * 1
        dy = dout * 1
        return dx, dy
```

```python
a1 = MulLayer
a1.forward(a1, 2, 3)
a1.backward(a1, 2, 3)
```

```python
import numpy as np
class ReLU:
    def __init__(self):
        self.mask = None
    
    def forward(self, x):
        self.mask = x<=0
        out = x.copy()
        out[self.mask] = 0

        return out
    def backward(self, dout):
        dout[self.mask] = 0
        dx = dout

        return dx
    
r1 = ReLU
r1.forward(r1,np.array([1,2,3,4,5,-9]))
```
